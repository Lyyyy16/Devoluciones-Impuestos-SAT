##Análisis 1: Describir la distribución
Usamos las medidas estádisticas para encontrar las distribuciones monetarias de nuestrso registros.

```python
# Lista de impuestos a analizar
impuestos = ['isr', 'ietu', 'iva', 'ieps', 'istuv', 'isan', 'ice', 'ide']

# Calculamos las estadísticas descriptivas para cada uno de ellos
resumen_impuestos = df[impuestos].agg(['mean', 'median', 'min', 'max']).transpose()

# Renombramos las columnas para que tengan un formato claro en español
resumen_impuestos.columns = ['Media', 'Mediana', 'Mínimo', 'Máximo']
resumen_impuestos.index.name = 'Impuesto'

display(resumen_impuestos)
```

```python
otros_impuestos = df['otros_impuestos']

media = float(otros_impuestos.mean())
mediana = float(otros_impuestos.median())
minimo = float(otros_impuestos.min())
maximo = float(otros_impuestos.max())

centro_extremos = pd.DataFrame({
    'Medida': ['Media', 'Mediana', 'Mínimo (Extremo Inferior)', 'Máximo (Extremo Superior)'],
    'Valor (Otros Impuestos)': [media, mediana, minimo, maximo]
})
display(centro_extremos)
```

##Cuantiles y dispersión

```python
# Lista de impuestos para el análisis de dispersión
impuestos_analisis = ['isr', 'ietu', 'iva', 'ieps', 'istuv', 'isan', 'ice', 'ide']

# Cálculo de métricas de dispersión por impuesto
cuantiles = df[impuestos_analisis].quantile([0.25, 0.50, 0.75])

q1 = cuantiles.loc[0.25]
mediana = cuantiles.loc[0.50]
q3 = cuantiles.loc[0.75]
iqr = q3 - q1
minimo = df[impuestos_analisis].min()
maximo = df[impuestos_analisis].max()
rango = maximo - minimo
desviacion = df[impuestos_analisis].std(ddof=0)

# Estructuración de los resultados en un DataFrame de dispersión
dispersion = pd.DataFrame({
    'Q1': q1,
    'Mediana (Q2)': mediana,
    'Q3': q3,
    'Rango': rango,
    'IQR': iqr,
    'Desviación Estándar (ddof=0)': desviacion
})

display(dispersion)
```
